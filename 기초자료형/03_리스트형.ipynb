```python
# 리스트 연산

# 1. 리스트 더하기
a = [1, 2, 3]
b = [4, 5, 6]
a+b

# 2. 리스트 곱하기(반복)
a = [1, 2, 3]
print(a * 3)
```

```python
# 리스트의 수정과 삭제

# 1. 리스트의 값 수정하기
a = [1, 2, 3]
a[2] = 4
print(a)

# 2. 리스트의 요소 삭제하기
a = [1, 2, 3]
del a[1]
print(a)

a = [1, 2, 3, 4, 5]
del a[2:]
print(a)
```

```python
# 리시트 관련 함수

# 1. 리스트에 요소 추가하기 append
a = [1, 2, 3]
a.append(4)
print(a)				# a.의미 = a라는 변수 안으로 들어간다
a.append([5, 6])
print(a)
```

```python
# 2. 리스트 정렬

# 2-1 sort(오름차순)
a = [1, 4, 3, 2]
a.sort()
print(a)

# 2-2 내림차순 sort"()"안에 reverse 값을 True로
a.sort(reverse=True)
print(a)
```

```python
# 3. 리스트 뒤집기 - reverse(리스트안에 값을 반대로 출력-뒤에서 부터 읽음)
a = ["a", "c", "b"]
# a.sort()
# print(a)

a.reverse()
print(a)
```

```python
# 4. 인덱스 번호 - index
a = [5, 4, 3, 2, 1]
print(a.index(3))
print(a.index(2))
```

```python
# 5. 리스트에 요소 삽입 - insert
a = [1, 2, 3]
a.insert(0, 4)
a
```

```python
# 6. 리스트요소 제거 - remove
a = [1, 2, 3, 1, 2, 3]
a.remove(3)
a
```

```python
# 7. 리스트 요소 꺼내기 - pop ()안에 기본 값은 -1
a = [1, 2, 3]
result1 = a.pop()
print(a)
print(result1)
```

```python
# 8. 요소의 개수 세기 - count
a = [1, 2, 3, 1]
a.count(1)
```

```python
# 9. 리스트 확장 - extend
a = [1, 2, 3]
a.extend([4, 5])
print(a)
```
