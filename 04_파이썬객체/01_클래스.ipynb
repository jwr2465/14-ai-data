```python
class Calculator:
	def __init__(self):   # self는 기본값으로 반드시 들어가야된다
		self.result = 0
	
	def add(self, num):
		self.result = self.result + num
		return self.result
	
	def sub(self, num):
		self.result -= num
		return self.result
	
	def mul(self, num):
		self.result = self.result * num
		return self.result

	def div(self, num):
		self.result /= num
		return self.result

```

```python
cal1 = Calculator()
cal1.add(10)
cal1.sub(2)
cal1.mul(5)
cal1.div(4)
```

```python
# 객체에 연산할 숫자 지정하기

class FourCal:
	def setdata(self, first, second):
		self.first = first
		self.second = second
		

	def add(self):
		result = self.first + self.second
		return result
	
	def sub(self):
		result = self.first - self.second
		return result

	def mul(self):
		result = self.first * self.second
		return result
	
	def div(self):
		result = self.first / self.second
		return result
	
cal3 = FourCal()
cal3.setdata(4, 2)


print(cal3.first)
print(cal3.second)
print(cal3.add())
print(cal3.mul())
```

```python
class FourCal:
	def __init__(self, frist, second):
		self.frist = frist
		self.second = second

	def add(self):
		result = self.frist + self.second
		return result
	
cal4 = FourCal(10, 20)
print(cal4.frist)
print(cal4.second)

cal4.add()
```

```python
# 클래스의 상속

class MoreFourCal(FourCal):
	def pow(self):
		result = self.frist ** self.second
		return result
```

```python
cal6 = MoreFourCal(10, 2)
print(cal6.add())
print(cal6.pow())
```
