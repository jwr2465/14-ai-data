```python
# 1. 일반적인 함수
# 입력값이 있고, 반환값이 있다

def add(a, b):
	result = a + b
	return result

sum = add(3, 4)
print(sum)
```

```python
# 2. 입력값이 없는 함수
# 입력값은 없고, 반환값은 있다

def say():
	return "hi"

a = say()
print(a)
```

```python
# 3. 반환값이 없는 함수
# 입력값은 있고, 반환값은 없는 경우

def add(a, b):
	print(f"{a}, {b}의 합은 {a+b}입니다.")

sum = add(3, 4)
print(sum)
```

```python
# 4. 입력값과 반환값 모두 없는 함수

def say():
	print("hi")

a = say()
print(a)
```

```python
# 매개변수를 지정하여 호출하기

def sub(a, b):  ### 괄호안에 들어가는 값이 매개변수
	return a-b

result = sub(a=7, b=3)
print(result)

result2 = sub(b=7, a=3)
print(result2)

print("안녕", "바이", sep=" ") # sep, end 기본 값은 변수 사이 공백
```

```python
#입력값이 몇 개가 될지 모를때는 어떻게 할까?

def add_many(*args):
	result = 0
	for i in args:
		result = result + i
	return result

result = add_many(1, 2, 3, 4, 5)
print(result)
```

```python
def add_mul(choice, *args):
	if choice == "add":
		result = 0
		for i in args:
			result = result + 1

	elif choice == "mul":
		result = 1
		for i in args:
			result = result * 1
```

```python
result = add_mul("add", 1, 2, 3, 4, 5)
print(result)
```

```python
# 함수의 반환값은 언제나 하나이다.

def add_and_mul(a, b):
	return a+b, a*b   # {}생략 - 튜플형태로 저장
```

```python
result = add_and_mul(3, 4)
print(result)
```

```python
def add_and_mul(a, b):
	return a+b
	return a*b
```

```python
result = add_and_mul(3, 4)
print(result)	# 상단 2번째 return 값은 흐리게 적혀있음 -> 첫번째 리턴값 이후로 읽지 않음
```

```python
def say_nick(nick):
	if nick == "학생":
		return				# break의 형태로 동작
	print(f"너의 별명은 {nick}이다")

say_nick("선생")
say_nick("학생")
```

```python
# 매개변수에 초깃값 미리 설정하기

def say_myself(name, age, man=True):
	print(f"너의 이름은 {name}이다")
	print(f"너의 나이는 {age}이다")
	if man:
		print("남자입니다")
	else:
		print("여자입니다.")

say_myself("홍길동", "25")
say_myself("김영희", 30, False)
```

```python
# 함수안에서 선언한 변수의 효력 범위

a = 1
def vartest(a):		# 함수안에서만 적용된다 = 지역변수
	a = a+1

vartest(a)
print(a)
```

```python
a = 1
def vartest():		
	global a		# 전역변수로 선언, 오류나기 쉬움, 함수안에서 선언한 함수는 그 안에서 사용하는게 좋다
	a = a+1

vartest()
print(a)
```

```python
# lambda 예약어
# 함수_이름 = lambda 매개변수1, 매개변수2, ... : 수행할 문장 <- 구조

def add(a, b):
	return a+b

add = lambda a, b: a+b

```

```python
# 함수의 독스트링(Docstring)	<--- 함수를 설명하는 명령어

def add(a, b):
	"""
	
	두 숫자를 더하는 함수

	Parameters:
	a(int, float): 첫 번째 숫자
	b(int, float): 두 번째 숫자

	return:
	int, float: 두 숫자의 합

	
	"""
	return a+b

print(add.__doc__)
```
