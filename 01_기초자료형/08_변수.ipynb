```python
a = [1, 2, 3]
print(id(a))

b = [1, 2, 3]
print(id(b))

a.append(4)
print(id(a))
```

```python
a = [1, 2, 3]
b = a[:] # <--- 값을 복제하는 개념

print(id(a))
print(id(b))
```

```python
# 변수를 만드는 여러가지 방법
a, b = ["python", "like"]   #구조분해, 언팩 의미
print(a)
print(b)

print(type(a))
```

```python
a, b = "python", "like"
print(a)
print(b)

print(type(a))
```

```python
a = b = "python"
print(a)
print(b)
print(type(a))
```
