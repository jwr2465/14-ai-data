```python
# 딕셔너리는 어떻게 만들까
dic = {"name" : "ken", "phone":"010-1234-1234", "birth":"01-01"}		#콜론으로 키와 밸류값 구분(앞에 값=키 / 뒤에 값=밸류)
print(type(dic))
```

```python
# 딕셔너리 쌍을 추가하기 ---- 리스트는 인덱스를 통해서 변경, 딕셔너리에서는 키값을 통해 변경
a = {1: "a"}
a[2]="b"
print(a)
a["name"] = "ken"
a
```

```python
# 딕셔너리 요소 삭제하기
print(a)
del a[1]
print(a)

del a["name"]
print(a)
```

```python
# 딕셔너리에 관련된 함수

# 1. key리스트 만들기 - keys
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print(profile.keys())

# 2. value 리스트 만들기 - values
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print(profile.values())

# 3. key, value 쌍으로 얻기 - itmes
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print(profile.items())

# 4. key와 value 쌍으로 모두 지우기 - clear
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print(profile.clear())

# 5. key로 value 얻기 - get
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print(profile.get("이름")) #profile의 ["이름"]과 같은 값


# 해당 key가 딕셔너리 안에 있는지 조사하기 - in
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
print("이름" in profile)

# key로 value 얻기 - pop
profile = {
	"이름" : "홍길동",
	"나이" : 25,
	"번호" : "010-1234-1234"
}
temp = profile.pop("이름")
print(temp)
print(profile)
```
