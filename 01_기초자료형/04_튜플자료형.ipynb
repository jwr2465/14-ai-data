```python
# 튜플은 어떻게 만들까?

a = ()
b = (1,)  #하나의 요소를 적을 때에는 뒤에 ","붙여넣기
c = (1, 2, 3)
d = 1, 2, 3		# 괄호 생략가능
e = ("a", "b", ("ab," "cd"))
f = 1,
g = 1


print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
```

```python
# 튜플 요소값을 삭제할 떄
a = (1, 2, "a", "b")
# del a[0] ---- 동작안함
```

```python
# 튜플 요소값을 변경하려 할때
a = (1, 2, "a", "b")
# a[0] = "c" ---- 튜플은 변경 및 삭제 모두 동작하지 않음
```
