```python
print(1, bool(1))
print(0, bool(0))
print(2, bool(2))
print("python", bool("ptthon"))
print("", bool(""))
print("[]", bool([]))
print("[1, 2, 3]", bool([1, 2, 3]))   #어떤 요소든 들어있으면 true

```

```python
if [1, 2, 3]:      #if 옆에 들어가는 값 = 로지컬 테스트
	print("참")
else:
	print("거짓")
```

```python
# 논리연산자

#1. and연산자
print(True and True)
print(True and False)
print(False and False)
print(True and True and False)	#False 값이 하나라도 있으면 False


#2. or연산자
print(True or True)
print(True or False)
print(False or False)


#3. not 연산자
print(not True)
print(not False)
print(not [])
```
